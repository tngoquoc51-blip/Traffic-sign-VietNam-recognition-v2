Notebook này dùng để kiểm tra nhanh dữ liệu: số lượng ảnh, phân bố các lớp, xem thử vài ảnh mẫu.

```python
import os
import sys
sys.path.append(os.path.join(os.getcwd(), '..', 'src'))

import config
from utils import count_images, load_class_names

for split in ['train', 'val', 'test']:
    img_dir = os.path.join(config.ROOT_DIR, 'data', split, 'images')
    print(f'{split}: {count_images(img_dir)} ảnh')
```

```python
names = load_class_names(config.DATA_YAML)
print('Các lớp biển báo:', names)
```

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import random

train_img_dir = os.path.join(config.ROOT_DIR, 'data', 'train', 'images')
if os.path.isdir(train_img_dir) and len(os.listdir(train_img_dir)) > 0:
    sample_files = random.sample(os.listdir(train_img_dir), min(4, len(os.listdir(train_img_dir))))
    fig, axes = plt.subplots(1, len(sample_files), figsize=(15, 5))
    if len(sample_files) == 1:
        axes = [axes]
    for ax, fname in zip(axes, sample_files):
        img = mpimg.imread(os.path.join(train_img_dir, fname))
        ax.imshow(img)
        ax.set_title(fname)
        ax.axis('off')
    plt.show()
else:
    print('Chưa có ảnh trong data/train/images. Hãy chạy src/data_preprocessing.py trước.')
```
